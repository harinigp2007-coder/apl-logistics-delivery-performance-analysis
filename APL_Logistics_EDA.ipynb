```python

import pandas as pd
import numpy as np

# Load the APL Logistics dataset
df = pd.read_csv("../data/APL_Logistics.csv", encoding="cp1252")

# Display basic information
print("Dataset Shape:", df.shape)

print("\nFirst 5 Records:")
display(df.head())

print("\nDataset Information:")
df.info()
```

```python

# Check missing values
print("Missing Values:")
display(df.isnull().sum()[df.isnull().sum() > 0])

# Check duplicate records
print("\nDuplicate Records:")
print(df.duplicated().sum())

# Check data types
print("\nColumn Data Types:")
display(df.dtypes)

# Statistical summary
print("\nStatistical Summary:")
display(df.describe())
```

```python

# Check shipping duration columns
shipping_cols = [
    "Days for shipping (real)",
    "Days for shipment (scheduled)"
]

print("Shipping Duration Data Types:")
display(df[shipping_cols].dtypes)

print("\nShipping Duration Summary:")
display(df[shipping_cols].describe())

print("\nMissing Shipping Duration Values:")
display(df[shipping_cols].isnull().sum())

print("\nInvalid or Zero Shipping Durations:")
display((df[shipping_cols] <= 0).sum())
```

```python

# Inspect records with zero shipping duration

print("Actual Shipping Days = 0")
display(
    df[df["Days for shipping (real)"] == 0][
        ["Days for shipping (real)",
         "Days for shipment (scheduled)",
         "Delivery Status",
         "Late_delivery_risk",
         "Shipping Mode"]
    ].head(10)
)

print("Scheduled Shipping Days = 0")
display(
    df[df["Days for shipment (scheduled)"] == 0][
        ["Days for shipping (real)",
         "Days for shipment (scheduled)",
         "Delivery Status",
         "Late_delivery_risk",
         "Shipping Mode"]
    ].head(10)
)
```

```python

# Check records where both shipping durations are zero

both_zero = df[
    (df["Days for shipping (real)"] == 0) &
    (df["Days for shipment (scheduled)"] == 0)
]

print("Both Actual and Scheduled Days are Zero:")
print("Count:", len(both_zero))

print("\nDelivery Status:")
display(both_zero["Delivery Status"].value_counts())

print("\nLate Delivery Risk:")
display(both_zero["Late_delivery_risk"].value_counts())

print("\nSample Records:")
display(both_zero[
    ["Days for shipping (real)",
     "Days for shipment (scheduled)",
     "Delivery Status",
     "Late_delivery_risk",
     "Shipping Mode"]
].head(10))
```

```python

scheduled_zero = df[
    (df["Days for shipment (scheduled)"] == 0) &
    (df["Days for shipping (real)"] > 0)
]

print("Scheduled Zero but Actual Days Greater than Zero:")
print("Count:", len(scheduled_zero))

print("\nDelivery Status:")
display(scheduled_zero["Delivery Status"].value_counts())

print("\nLate Delivery Risk:")
display(scheduled_zero["Late_delivery_risk"].value_counts())

print("\nSample Records:")
display(scheduled_zero[
    ["Days for shipping (real)",
     "Days for shipment (scheduled)",
     "Delivery Status",
     "Late_delivery_risk",
     "Shipping Mode"]
].head(10))
```

```python

# Check negative shipping durations

print("Negative Actual Shipping Days:")
print((df["Days for shipping (real)"] < 0).sum())

print("\nNegative Scheduled Shipping Days:")
print((df["Days for shipment (scheduled)"] < 0).sum())

# Check maximum shipping durations
print("\nMaximum Actual Shipping Days:")
print(df["Days for shipping (real)"].max())

print("\nMaximum Scheduled Shipping Days:")
print(df["Days for shipment (scheduled)"].max())
```

```python

# Calculate Delivery Delay Gap

df["Delivery Delay Gap"] = (
    df["Days for shipping (real)"] -
    df["Days for shipment (scheduled)"]
)

# Classify delivery performance
df["Delivery Performance"] = np.select(
    [
        df["Delivery Delay Gap"] > 0,
        df["Delivery Delay Gap"] == 0,
        df["Delivery Delay Gap"] < 0
    ],
    [
        "Delayed",
        "On-time",
        "Early"
    ],
    default="Unknown"
)

# Display results
print("Delivery Performance Distribution:")
display(df["Delivery Performance"].value_counts())

print("\nDelay Gap Summary:")
display(df["Delivery Delay Gap"].describe())

print("\nSample Records:")
display(df[
    ["Days for shipping (real)",
     "Days for shipment (scheduled)",
     "Delivery Delay Gap",
     "Delivery Performance",
     "Delivery Status",
     "Late_delivery_risk"]
].head(10))
```

```python

# Create a separate dataset for delivery analysis

analysis_df = df[
    (df["Delivery Status"] != "Shipping canceled") &
    (df["Days for shipment (scheduled)"] > 0)
].copy()

# Calculate delivery delay gap
analysis_df["Delivery Delay Gap"] = (
    analysis_df["Days for shipping (real)"] -
    analysis_df["Days for shipment (scheduled)"]
)

# Classify delivery performance
analysis_df["Delivery Performance"] = np.where(
    analysis_df["Delivery Delay Gap"] > 0,
    "Delayed",
    np.where(
        analysis_df["Delivery Delay Gap"] == 0,
        "On-time",
        "Early"
    )
)

# Display results
print("Original Dataset Shape:", df.shape)
print("Clean Analysis Dataset Shape:", analysis_df.shape)

print("\nDelivery Performance Distribution:")
display(analysis_df["Delivery Performance"].value_counts())

print("\nDelivery Status Distribution:")
display(analysis_df["Delivery Status"].value_counts())

print("\nDelay Gap Summary:")
display(analysis_df["Delivery Delay Gap"].describe())
```

```python

# Delivery Risk Analysis

print("Late Delivery Risk Distribution:")

display(analysis_df["Late_delivery_risk"].value_counts())

print("\nLate Delivery Risk Percentage:")

display(
    analysis_df["Late_delivery_risk"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python

# Compare Delivery Status with Late Delivery Risk

risk_comparison = pd.crosstab(
    analysis_df["Delivery Status"],
    analysis_df["Late_delivery_risk"]
)

print("Delivery Status vs Late Delivery Risk:")
display(risk_comparison)
```

```python

# Delivery Performance by Shipping Mode

shipping_analysis = analysis_df.groupby(
    "Shipping Mode"
)["Delivery Delay Gap"].mean().sort_values(ascending=False)

print("Average Delivery Delay by Shipping Mode:")
display(shipping_analysis)
```

```python

# Delivery Performance by Shipping Mode

shipping_performance = pd.crosstab(
    analysis_df["Shipping Mode"],
    analysis_df["Delivery Performance"]
)

print("Delivery Performance by Shipping Mode:")
display(shipping_performance)
```

```python

# Average Delivery Delay by Market

market_analysis = analysis_df.groupby(
    "Market"
)["Delivery Delay Gap"].mean().sort_values(ascending=False)

print("Average Delivery Delay by Market:")
display(market_analysis)
```

```python

# ============================================================
# APL LOGISTICS - COMPLETE REMAINING PROJECT ANALYSIS
# Unified Mentor Internship Project
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# ------------------------------------------------------------
# 1. PROJECT PATHS AND DATA PREPARATION
# ------------------------------------------------------------

PROJECT_DIR = Path("..")

OUTPUT_DIR = PROJECT_DIR / "outputs"
CHART_DIR = OUTPUT_DIR / "charts"
CLEAN_DIR = OUTPUT_DIR / "cleaned_data"
REPORT_DIR = OUTPUT_DIR / "reports"

for folder in [CHART_DIR, CLEAN_DIR, REPORT_DIR]:
    folder.mkdir(parents=True, exist_ok=True)

# Create clean analysis dataset
analysis_df = df[
    (df["Delivery Status"] != "Shipping canceled") &
    (df["Days for shipment (scheduled)"] > 0)
].copy()

analysis_df["Delivery Delay Gap"] = (
    analysis_df["Days for shipping (real)"] -
    analysis_df["Days for shipment (scheduled)"]
)

analysis_df["Delivery Performance"] = np.where(
    analysis_df["Delivery Delay Gap"] > 0,
    "Delayed",
    np.where(
        analysis_df["Delivery Delay Gap"] == 0,
        "On-time",
        "Early"
    )
)

# Save cleaned dataset
analysis_df.to_csv(
    CLEAN_DIR / "delivery_analysis_cleaned.csv",
    index=False
)

print("Clean Dataset Shape:", analysis_df.shape)


# ------------------------------------------------------------
# 2. ORDER REGION ANALYSIS
# ------------------------------------------------------------

region_analysis = analysis_df.groupby("Order Region").agg(
    Total_Shipments=("Delivery Delay Gap", "count"),
    Average_Delay=("Delivery Delay Gap", "mean"),
    Late_Risk_Count=("Late_delivery_risk", "sum")
)

region_analysis["Late_Risk_Percentage"] = (
    region_analysis["Late_Risk_Count"] /
    region_analysis["Total_Shipments"] * 100
)

region_analysis = region_analysis.round(2).sort_values(
    "Average_Delay", ascending=False
)

print("\nORDER REGION ANALYSIS")
display(region_analysis)

region_analysis.to_csv(
    OUTPUT_DIR / "region_analysis.csv"
)


# ------------------------------------------------------------
# 3. ORDER COUNTRY ANALYSIS
# ------------------------------------------------------------

country_analysis = analysis_df.groupby("Order Country").agg(
    Total_Shipments=("Delivery Delay Gap", "count"),
    Average_Delay=("Delivery Delay Gap", "mean"),
    Late_Risk_Count=("Late_delivery_risk", "sum")
)

country_analysis["Late_Risk_Percentage"] = (
    country_analysis["Late_Risk_Count"] /
    country_analysis["Total_Shipments"] * 100
)

country_analysis = country_analysis.round(2).sort_values(
    "Average_Delay", ascending=False
)

print("\nTOP 10 COUNTRIES BY AVERAGE DELIVERY DELAY")
display(country_analysis.head(10))

country_analysis.to_csv(
    OUTPUT_DIR / "country_analysis.csv"
)


# ------------------------------------------------------------
# 4. CUSTOMER SEGMENT IMPACT ANALYSIS
# ------------------------------------------------------------

segment_analysis = analysis_df.groupby("Customer Segment").agg(
    Total_Shipments=("Delivery Delay Gap", "count"),
    Average_Delay=("Delivery Delay Gap", "mean"),
    Delayed_Shipments=(
        "Delivery Performance",
        lambda x: (x == "Delayed").sum()
    ),
    On_Time_Shipments=(
        "Delivery Performance",
        lambda x: (x != "Delayed").sum()
    ),
    Late_Risk_Count=("Late_delivery_risk", "sum")
)

segment_analysis["Delay_Percentage"] = (
    segment_analysis["Delayed_Shipments"] /
    segment_analysis["Total_Shipments"] * 100
)

segment_analysis["Late_Risk_Percentage"] = (
    segment_analysis["Late_Risk_Count"] /
    segment_analysis["Total_Shipments"] * 100
)

segment_analysis = segment_analysis.round(2)

print("\nCUSTOMER SEGMENT ANALYSIS")
display(segment_analysis)

segment_analysis.to_csv(
    OUTPUT_DIR / "customer_segment_analysis.csv"
)


# ------------------------------------------------------------
# 5. KEY PERFORMANCE INDICATORS (KPIs)
# ------------------------------------------------------------

total_shipments = len(analysis_df)

delayed_count = (
    analysis_df["Delivery Performance"] == "Delayed"
).sum()

early_count = (
    analysis_df["Delivery Performance"] == "Early"
).sum()

on_time_count = (
    analysis_df["Delivery Performance"] == "On-time"
).sum()

on_time_delivery_rate = (
    (early_count + on_time_count) / total_shipments * 100
)

average_delivery_delay = (
    analysis_df["Delivery Delay Gap"].mean()
)

late_delivery_risk_ratio = (
    analysis_df["Late_delivery_risk"].mean() * 100
)

delayed_percentage = (
    delayed_count / total_shipments * 100
)

kpi_data = {
    "Total Eligible Shipments": total_shipments,
    "Delayed Shipments": delayed_count,
    "Early Shipments": early_count,
    "On-time Shipments": on_time_count,
    "On-Time Delivery Rate (%)": round(on_time_delivery_rate, 2),
    "Average Delivery Delay (Days)": round(average_delivery_delay, 2),
    "Late Delivery Risk Ratio (%)": round(late_delivery_risk_ratio, 2),
    "Delayed Shipment Percentage (%)": round(delayed_percentage, 2)
}

kpi_df = pd.DataFrame(
    list(kpi_data.items()),
    columns=["KPI", "Value"]
)

print("\nKEY PERFORMANCE INDICATORS")
display(kpi_df)

kpi_df.to_csv(
    OUTPUT_DIR / "project_kpis.csv",
    index=False
)


# ------------------------------------------------------------
# 6. SHIPPING MODE EFFICIENCY INDEX
# ------------------------------------------------------------

shipping_kpi = analysis_df.groupby("Shipping Mode").agg(
    Total_Shipments=("Delivery Delay Gap", "count"),
    Average_Delay=("Delivery Delay Gap", "mean"),
    Delayed_Count=(
        "Delivery Performance",
        lambda x: (x == "Delayed").sum()
    )
)

shipping_kpi["Delayed_Percentage"] = (
    shipping_kpi["Delayed_Count"] /
    shipping_kpi["Total_Shipments"] * 100
)

shipping_kpi = shipping_kpi.round(2)

print("\nSHIPPING MODE EFFICIENCY")
display(shipping_kpi)

shipping_kpi.to_csv(
    OUTPUT_DIR / "shipping_mode_efficiency.csv"
)


# ------------------------------------------------------------
# 7. REGIONAL DELAY INDEX
# ------------------------------------------------------------

regional_delay_index = (
    analysis_df.groupby("Order Region")["Delivery Delay Gap"]
    .mean()
    .round(2)
    .sort_values(ascending=False)
)

print("\nREGIONAL DELAY INDEX")
display(regional_delay_index)

regional_delay_index.to_csv(
    OUTPUT_DIR / "regional_delay_index.csv"
)


# ------------------------------------------------------------
# 8. VISUALIZATION FUNCTION
# ------------------------------------------------------------

def save_bar_chart(data, title, xlabel, ylabel, filename):

    plt.figure(figsize=(11, 6))

    data.plot(kind="bar")

    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()

    plt.savefig(
        CHART_DIR / filename,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()


# ------------------------------------------------------------
# 9. GENERATE PROJECT CHARTS
# ------------------------------------------------------------

save_bar_chart(
    analysis_df["Delivery Performance"].value_counts(),
    "Overall Delivery Performance",
    "Delivery Performance",
    "Number of Shipments",
    "delivery_performance.png"
)

save_bar_chart(
    analysis_df.groupby("Shipping Mode")[
        "Delivery Delay Gap"
    ].mean().sort_values(ascending=False),
    "Average Delay by Shipping Mode",
    "Shipping Mode",
    "Average Delay (Days)",
    "shipping_mode_delay.png"
)

save_bar_chart(
    region_analysis["Average_Delay"],
    "Average Delivery Delay by Order Region",
    "Order Region",
    "Average Delay (Days)",
    "region_delay.png"
)

save_bar_chart(
    analysis_df.groupby("Market")[
        "Delivery Delay Gap"
    ].mean().sort_values(ascending=False),
    "Average Delivery Delay by Market",
    "Market",
    "Average Delay (Days)",
    "market_delay.png"
)

save_bar_chart(
    segment_analysis["Average_Delay"],
    "Average Delivery Delay by Customer Segment",
    "Customer Segment",
    "Average Delay (Days)",
    "customer_segment_delay.png"
)

save_bar_chart(
    country_analysis.head(10)["Average_Delay"],
    "Top 10 Countries by Average Delivery Delay",
    "Order Country",
    "Average Delay (Days)",
    "top_10_country_delay.png"
)


# ------------------------------------------------------------
# 10. DELAY GAP HISTOGRAM
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.hist(
    analysis_df["Delivery Delay Gap"],
    bins=20,
    edgecolor="black"
)

plt.title("Delivery Delay Gap Distribution")
plt.xlabel("Delivery Delay Gap (Days)")
plt.ylabel("Number of Shipments")

plt.tight_layout()

plt.savefig(
    CHART_DIR / "delivery_delay_histogram.png",
    dpi=300
)

plt.show()
plt.close()


# ------------------------------------------------------------
# 11. GENERATE RESEARCH PAPER
# ------------------------------------------------------------

highest_region = region_analysis["Average_Delay"].idxmax()
highest_country = country_analysis["Average_Delay"].idxmax()
highest_segment = segment_analysis["Average_Delay"].idxmax()
highest_shipping = shipping_kpi["Average_Delay"].idxmax()

research_paper = f"""
# Delivery Performance, Delay Risk, and Logistics Efficiency
## Analysis in Global Supply Chain Operations

### 1. Introduction

This project analyzes delivery performance and logistics efficiency
using the APL Logistics supply chain dataset.

### 2. Problem Statement

Delivery delays can affect customer satisfaction, shipping costs,
and operational efficiency. This analysis evaluates delivery
performance, shipping modes, geographical regions, and customer
segments to identify potential logistics bottlenecks.

### 3. Dataset Description

Original dataset records: {len(df)}

Eligible records used for delivery analysis: {total_shipments}

The analysis excludes cancelled shipments and records with
scheduled shipping duration equal to zero.

### 4. Methodology

- Data cleaning and validation
- Delivery delay gap calculation
- Delivery performance classification
- Late delivery risk analysis
- Shipping mode comparison
- Market, region and country analysis
- Customer segment analysis
- KPI calculation and visualization

### 5. Key Performance Indicators

- On-Time Delivery Rate: {on_time_delivery_rate:.2f}%
- Average Delivery Delay Gap: {average_delivery_delay:.2f} days
- Late Delivery Risk Ratio: {late_delivery_risk_ratio:.2f}%
- Total Eligible Shipments: {total_shipments}

On-time delivery rate includes both early and exactly on-time
shipments.

### 6. Delivery Performance

Delayed shipments: {delayed_count}
Early shipments: {early_count}
On-time shipments: {on_time_count}

### 7. Shipping Mode Analysis

The shipping mode with the highest average delay gap is:
{highest_shipping}.

### 8. Regional Analysis

The region with the highest average delivery delay gap is:
{highest_region}.

### 9. Country Analysis

The country with the highest average delivery delay gap is:
{highest_country}.

### 10. Customer Segment Analysis

The customer segment with the highest average delivery delay gap is:
{highest_segment}.

### 11. Limitations

The late_delivery_risk field is an existing dataset label.
It matches the delivery status classification in this analysis.
Therefore, it is treated as an existing risk indicator and not
as an independently validated predictive model.

The analysis is observational and does not establish the causes
of delivery delays.

### 12. Recommendations

- Monitor shipping modes with higher average delay gaps.
- Review operational bottlenecks in high-delay regions.
- Monitor country-level delivery performance.
- Compare delivery experience across customer segments.
- Use dashboard filters for continuous performance monitoring.

### 13. Conclusion

This project provides a data-driven overview of delivery
performance, delay risk, shipping mode efficiency, geographical
patterns, and customer segment impact.

The findings can support logistics teams in identifying
areas that require further operational investigation.
"""

(REPORT_DIR / "Research_Paper.md").write_text(
    research_paper,
    encoding="utf-8"
)


# ------------------------------------------------------------
# 12. GENERATE EXECUTIVE SUMMARY
# ------------------------------------------------------------

executive_summary = f"""
# EXECUTIVE SUMMARY

## Project
Delivery Performance, Delay Risk, and Logistics Efficiency
Analysis in Global Supply Chain Operations

## Organization Context
APL Logistics (KWE Group)

## Project Overview

This project evaluates delivery performance using
{total_shipments:,} eligible shipment records.

## Main Findings

- On-Time Delivery Rate: {on_time_delivery_rate:.2f}%
- Average Delivery Delay Gap: {average_delivery_delay:.2f} days
- Late Delivery Risk Ratio: {late_delivery_risk_ratio:.2f}%
- Delayed Shipments: {delayed_count:,}
- Early Shipments: {early_count:,}
- On-Time Shipments: {on_time_count:,}

## Areas Requiring Attention

- Shipping Mode: {highest_shipping}
- Order Region: {highest_region}
- Order Country: {highest_country}
- Customer Segment: {highest_segment}

These are the groups with the highest observed average
delivery delay gap in their respective categories.

## Recommendations

1. Monitor delayed shipment patterns.
2. Review shipping mode efficiency.
3. Investigate high-delay geographical areas.
4. Evaluate delivery performance across customer segments.
5. Use dashboard filters for operational monitoring.

## Conclusion

The analysis provides insights into delivery performance
and logistics efficiency and supports data-driven
operational decision-making.

Note: The analysis identifies patterns and associations.
It does not independently establish the causes of delays
or provide a validated predictive model.
"""

(REPORT_DIR / "Executive_Summary.md").write_text(
    executive_summary,
    encoding="utf-8"
)


# ------------------------------------------------------------
# 13. FINAL OUTPUT SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 55)
print("PROJECT ANALYSIS COMPLETED SUCCESSFULLY")
print("=" * 55)

print("\nGenerated files:")

for file in OUTPUT_DIR.rglob("*"):
    if file.is_file():
        print(file)

print("\nAll remaining analysis and report files generated!")
```
